deciding mean or median

```python
for column in diabetes_columns:
  skew_value = diabetes[column].skew()


  if -0.5 <= skew_value <= 0.5:
    print(column, "-> Use mean")
    diabetes[column] = diabetes[column].fillna(diabetes[column].mean())
  else:
    print(column, "-> Use median")
    diabetes[column] = diabetes[column].fillna(diabetes[column].median())
```

```python
from sklearn.model_selection import train_test_split
X = diabetes[diabetes_input]
y = diabetes[diabetes_target]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42,stratify=y)
print("Training rows:", X_train.shape[0])
print("Testing rows:", X_test.shape[0])
```

```python
from sklearn. tree import DecisionTreeClassifier

decision_tree = DecisionTreeClassifier(
  max_depth=4,
  class_weight="balanced",
  random_state=42

)
```

```python
decision_tree.fit(X_train, y_train)
print("Decision Tree Model train successfully.")
```

```python
diabetes_predictions = decision_tree.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
accuracy = accuracy_score(y_test, diabetes_predictions)
print("Accuracy:", accuracy)


print("Classification Report:")
print(classification_report(y_test, diabetes_predictions))


print("Confusion Matrix:")
print(confusion_matrix(y_test, diabetes_predictions))
```
